Indexers: loc, iloc, and ix

```python
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
```

```python
print(data[3])
print(data[1:3]) #1:3 uses python indexing
```

```python
print(data.loc[1:3]) #refernces the expilictly named index
print(data.iloc[1:3]) #refernces the python index
```

Dataframe

```python
#Dataframe = chart
area = pd.Series({'California': 423967, 'Texas': 695662,
                  'New York': 141297, 'Florida': 170312,
                  'Illinois': 149995})
pop = pd.Series({'California': 38332521, 'Texas': 26448193,
                 'New York': 19651127, 'Florida': 19552860,
                 'Illinois': 12882135})
data = pd.DataFrame({'area':area, 'pop':pop})
data
```

```python
data['area']
data.area #same thing but doesnt always work
```

```python
#We can add stuff too
data['density'] = data['pop'] / data['area']
data
```

```python
#Dataframe can be a 2d array
print(data.values)
print(data.values[0]) #can access rows
```

```python
data[data.density > 100]
```
