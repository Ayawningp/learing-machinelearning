```python
def find_reciporcals(values):
    output = np.zeros(len(values))
    for i in range(len(values)):
        output[i] = 1/values[i]
    return output

values = np.random.randint(1,10,size=10)
print(find_reciporcals(values))
print(1/values)
```

```python
xnv = values.reshape(2,5)
print(xnv **2)
```

```python
x = np.arange(1,10)
print(np.add.reduce(x)) #repeatdly adds numbers in the array until one value remains
print(np.multiply.reduce(x)) #repeatdly multiplies
print(np.add.accumulate(x)) #shows intermediate steps
print(np.sum(x))
```

```python
plotting k nearest neighbors
```

```python
x = np.random.rand(10,2)
print(x)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.scatter(x[:,0], x[:,1]);
#print(x)

#distance formula between each point 
diff_2 = (x[:,np.newaxis,:] - x[np.newaxis, :, :])**2 #logic expand into 3d array so entries inside are the distance
dist = diff_2.sum(-1) #sums along 3rd axis
#print(dist)

#nearest k neighbors
k= 2
k_near = np.argpartition(dist, k+1) #index closest to the point
#print(k_near)

#line between nears k neighbors
for i in range(x.shape[0]):
    for j in k_near[i, :k+1]:
        plt.plot(*zip(x[i], x[j]), color='black') #draws a line closest to k points
```

structred arrays (data of multiple categories)

```python
import numpy as np
name = ['Alice', 'Bob', 'Cathy', 'Doug']
age = [25, 45, 37, 19]
weight = [55.0, 85.5, 68.0, 61.5]
```

```python
data = np.zeros(4, dtype={'names':('name','age','weight'), 'formats':('U10', 'i4', 'f8')}) 
#4 correlates to number of data points
print(data.dtype)
#U10 string of len10    i4 32bit int    f8 64bit int
```

```python
data['name'] = name
data['age'] = age
data['weight'] = weight
print(data)
print(data[0]) #information of 1st index
print(data[-1]['age']) #age of last index
print(data[data['age']<30]['name']) #names of people under 30
```
