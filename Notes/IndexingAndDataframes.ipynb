Python Missing Data & More indexing

```python
import numpy as np
import pandas as pd
```

```python
#python uses None to represent arbitary or missing values
vals1 = np.array([1,None,3,4]) #None is an object type
vals1
```

```python
#numpy uses NaN (not a number)
vals2 = np.array([1,np.nan,3,4]) #float
vals2
```

```python
#when doing arthemtic with NaN the solution will be Nan
print(1+np.nan)
print(1* np.nan)
```

```python
#Numpy can ignore NaN
np.nansum(vals2) #nanmin, nanmax
```

In Pandas

For dataframes

```python
df = pd.DataFrame([[1,      np.nan, 2],
                   [2,      3,      5],
                   [np.nan, 4,      6]])
df
```

```python
df.dropna() #dropna will default remove rows with null values
```

```python
#Instead of dropping we can fill with a set value
df.fillna(0)
```

```python
df.fillna(method='ffill')
```

Hierarchial Indexing

```python
index = [('California', 2000), ('California', 2010),
         ('New York', 2000), ('New York', 2010),
         ('Texas', 2000), ('Texas', 2010)]
populations = [33871648, 37253956,
               18976457, 19378102,
               20851820, 25145561]
```

```python
pop = pd.Series(populations, index=index)
pop #using population as keys
```

```python
#Multi indexing gives us multiple labels for the same value
index = pd.MultiIndex.from_tuples(index)
index
```

```python
pop = pop.reindex(index)
pop #Hierarchial representation
```

```python
print(pop[:, 2010]) #all data in 2010
print(pop['California']) #all the data from California
```

```python
#unstack will convert a series into a data frame - stack is the opposite
pop_df = pop.unstack()
pop_df
```

```python
#With a multi index we can easily add another column
pop_df.stack
pop_df = pd.DataFrame({'Total' : pop,
                       'Under18': [9267089, 9284094,
                                   4687374, 4318033,
                                   5906301, 6879014]})
pop_df
```

```python
#Fraction of people under 18
f_u18 = pop_df['Under18'] / pop_df['Total']
f_u18.unstack()
```

```python
import pandas as pd
import numpy as np
import numexpr

class display(object):
    """Display HTML representation of multiple objects"""
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    def __init__(self, *args):
        self.args = args
        
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                         for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                           for a in self.args)
```

```python
pop = pd.read_csv('state-population.csv')
areas = pd.read_csv('state-areas.csv')
abbrevs = pd.read_csv('state-abbrevs.csv')

display('pop.head()', 'areas.head()', 'abbrevs.head()') #head returns n rows
```

```python
#rank US states by population density

#Start by merging the pop and abbreviation and then drop the abbrev so no duplicate
merged = pd.merge(pop, abbrevs, how='outer', left_on='state/region', right_on='abbreviation')
merged = merged.drop('abbreviation',axis=1)
merged.head()
```

```python
#check for mismatches
merged.isnull().any()
```

```python
#Puerto Rico doesn't have population data or state entery
merged[merged['population'].isnull()].head()
```

```python
#Finding which states are null
merged.loc[merged['state'].isnull(), 'state/region'].unique()
```

```python
#Adding states to the abbreviations
merged.loc[merged['state/region'] == 'PR', 'state'] = 'Puerto Rico'
merged.loc[merged['state/region'] == 'USA', 'state'] = 'United States'
merged.isnull().any()
```

```python
#The states are good so we can merge the areas now cause they share similar indexes
final = pd.merge(merged, areas, on='state', how='left')
final.head()
```

```python
#check for nulls
final.isnull().any()
```

```python
#Find which state have null area
final['state'][final['area (sq. mi)'].isnull()].unique()#USA
final.dropna(inplace=True)
final.head()
```

```python
#find the 2010 year
data2010 = final.query("year == 2010 & ages == 'total'")
data2010.head()
```

```python
data2010.set_index('state' , inplace=True) #set our index to state
data2010.head()
```

```python
density = data2010['population'] / data2010['area (sq. mi)']
density.sort_values(ascending= False).head()
```
