```python
import numpy as np
import pandas as pd

#import data
pop = pd.read_csv('state-population.csv')
area = pd.read_csv('state-areas.csv')
abbrev = pd.read_csv('state-abbrevs.csv')
```

```python
merged = pd.merge(area, abbrev, how='outer')

merged[merged['abbreviation'].isnull()].head()
```

```python
#ranking population density from the year 2010
merged = pd.merge(area, abbrev, how='outer')

#Finding Null Values
merged.isnull().any()
merged[merged['abbreviation'].isnull()].head() #Puerto Rico has no abbreviation
merged.loc[merged['state'] == 'Puerto Rico' , 'abbreviation' ] = 'PR'

#merging population and merged
final = pd.merge(pop, merged, how='outer', left_on='state/region', right_on='abbreviation')
final.drop('abbreviation', axis=1, inplace=True) #drop the duplicate abbreviation
final.dropna(inplace=True) #Drop USA values cause not a state
#final.head()
```

```python
data2010 = final.query("year == 2010 & ages == 'total'")
data2010.set_index('state', inplace=True)
density = data2010['population'] / data2010['area (sq. mi)']
density.sort_values(ascending=False, inplace= True)
density.head()
```

Cleaning Titantic dataset

```python
import seaborn as sns
import pandas as pd
df = sns.load_dataset('titanic')

#df.info()
#df.isnull().any()
#missing age, embarked, deck, and embarked town

#dropping data
#input age with median age of passengar class
df['age'] = df['age'].fillna(df.groupby('pclass')['age'].transform('median'))
df = df.drop('deck', axis=1) #most of deck is missing so might as well drop
df = df.dropna(subset=['embarked', 'embark_town']) #subset targets specific rows

#convert categorical data to numeric values
df['sex'] = df['sex'].map({'male' : 1 , 'female' : 0})
df['class'] = df['class'].map({'Third' : 3, 'Second' : 2, 'First':1})

#drop redunant columns
df = df.drop(columns=['adult_male', 'who', 'alive', 'alone', 'embarked'], axis=1)
df
```
