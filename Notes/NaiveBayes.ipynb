Gaussian Naives Bayes Example

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
```

```python
#I did this by hand
from sklearn.naive_bayes import GaussianNB

#Training data for heights
#Male(M, 0): 6.0, 5.8, 6.2: Mean = 6.0, Variance = 0.0267
#Female(F, 1):5.0, 5.4, 5.2: mean = 6.0, Variance = 0.0267

#prepare data
X = np.array([[6.0], [5.8], [6.2], [5.0], [5.4], [5.2]])
y = np.array([0, 0, 0, 1, 1, 1,]) #0 for male, 1 for female

#prepare model
model = GaussianNB()
model.fit(X, y)

#predict new person
X_new = np.array([[5.4]])
prediction = model.predict(X_new)
probabilities = model.predict_proba(X_new)

#results
gender_mapping = {0: "Male", 1: "Female"}
print(f"Predicted Class: {gender_mapping[prediction[0]]}")
print(f"Probability Distribution (Male, Female): {probabilities[0]}")
```

Practice on Sklearn(Iris)

```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

#load data set
iris = datasets.load_iris()
X = iris.data
y= iris.target

#Training and Testing
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, train_size=0.2)

#prepare model
model = GaussianNB()
model.fit(Xtrain, ytrain)

#predict label
ypred = model.predict(Xtest)

print(f"Accuracy Score: {accuracy_score(ytest, ypred):.4f}")
print("\nClassification Report:\n", classification_report(ytest, ypred, target_names=iris.target_names))
```

```python
flower = np.array([[3.4, 2.4, 1, 0.3]])
prediction = model.predict(flower)
species = iris.target_names[prediction[0]]
probabilities = model.predict_log_proba(flower)

print(f"Predicted Species Name : {species}")
print(f"Probability Distribution : {probabilities[0]}")
```

Breast Cancer

```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

#load data set
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

#Training and Testing
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, train_size=0.2)

#Prepare model
model = GaussianNB()
model.fit(Xtrain, ytrain)

#predict label
ypred = model.predict(Xtest)

print(f"Accuracy Score: {accuracy_score(ytest, ypred):.4f}")
print("\nClassification Report:\n", classification_report(ytest, ypred, target_names=breast_cancer.target_names))
```
