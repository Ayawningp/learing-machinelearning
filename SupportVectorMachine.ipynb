Support Vector Machine 

```python
import numpy as np
from sklearn.svm import SVC

#load data
X = np.array([[1,2], [2,1], [0,0]]) #points handwritten
y = np.array([1,1,-1]) #values

model = SVC(kernel='linear', C=1e5) 
#C is how strict the model is small is high error tolerance large is low error tolerance(overfitting)
model.fit(X,y)

#find weights and bias
weights = model.coef_[0]
bias = model.intercept_[0]

print("--- PRIMAL SOLUTION ---")
print(f"Weight Vector (w): {weights}")
print(f"Bias Scalar   (b): {bias:.4f}")
print(f"Simplified Line  : {weights[0]:.2f}x1 + {weights[1]:.2f}x2 + ({bias:.2f}) = 0\n")
```

```python
import numpy as np
from sklearn.svm import SVC
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
#ML models are biased towards features with large scalers import for distance based machine learning models
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

#load data
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

#spilt data into training and test
Xtrain, Xtest, ytrain, ytest = train_test_split(X,y, train_size=0.2)
scaler = StandardScaler()
Xtrainscaled = scaler.fit_transform(Xtrain) #calculates parameters(mean and varience)
Xtestscaled = scaler.transform(Xtest) #applies parameters

#train model
model = SVC(kernel='rbf', C=1)
model.fit(Xtrainscaled, ytrain)

#evaluate
ypred = model.predict(Xtestscaled)
print("Accuracy:", accuracy_score(ytest, ypred))
```
