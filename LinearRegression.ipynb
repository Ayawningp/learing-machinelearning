Linear Regression

```python
import numpy as np
from sklearn.linear_model import LinearRegression

#dataset 
X = [[1], [2], [3]]
y = [2,3,5]

#prepare model
model = LinearRegression()
model.fit(X,y)

#Constants
print(f"Slope : {model.coef_[0]}")
print(f'Intercept : {model.intercept_}')
```

```python
new_data = np.array([[4]])
print(f"Predict : {model.predict(new_data)}")
```
